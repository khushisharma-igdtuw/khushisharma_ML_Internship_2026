natural numbers and math function

```python
# 1. Print first 10 natural numbers
def print_natural_numbers():
    print("First 10 natural numbers:")
    for i in range(1, 11):
        print(i, end=" ")
    print()

# 2. Sum of first N natural numbers
def sum_of_n_numbers(n):
    return n * (n + 1) // 2

# 3. Reverse a number
def reverse_number(num):
    return int(str(num)[::-1])

# 4. Count digits in a number
def count_digits(num):
    return len(str(abs(num)))

# 5. Check palindrome number
def is_palindrome(num):
    return str(num) == str(num)[::-1]

# 6. Generate Fibonacci series
def fibonacci(n):
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib[:n]

print_natural_numbers()
print("Sum of first 10 natural numbers:", sum_of_n_numbers(10))
print("Reverse of 12345:", reverse_number(12345))
print("Digits in 98765:", count_digits(98765))
print("Is 121 palindrome?", is_palindrome(121))
print("First 10 Fibonacci numbers:", fibonacci(10))
```

calculator using functions

```python
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b): return a / b if b != 0 else "Error: Division by zero"

def calculator():
    print("\n--- Calculator ---")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = input("Enter your choice (1-4): ")
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    operations = {'1': add, '2': subtract, '3': multiply, '4': divide}

    if choice in operations:
        result = operations[choice](a, b)
        print(f"Result: {result}")
    else:
        print("Invalid choice!")

calculator()
```

student details

```python
# Create a text file and store student details
def create_student_file():
    with open("students.txt", "w") as f:
        f.write("Name, Marks\n")
        f.write("Alice, 85\n")
        f.write("Bob, 90\n")
        f.write("Charlie, 78\n")
    print("Student file created successfully!")

# Read data from a file
def read_student_file():
    print("\n--- Student Data ---")
    with open("students.txt", "r") as f:
        for line in f:
            print(line.strip())

create_student_file()
read_student_file()
```

division by zero

```python
def safe_divide():
    try:
        a = float(input("\nEnter numerator: "))
        b = float(input("Enter denominator: "))
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
    except ValueError:
        print("Error: Please enter valid numbers!")
    finally:
        print("Division operation attempted.")

safe_divide()
```

a class of students with name and marks

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"\nStudent Name: {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Grade: {self.get_grade()}")

    def get_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        else:
            return "F"

# Create student objects
s1 = Student("Alice", 92)
s2 = Student("Bob", 75)
s3 = Student("Charlie", 55)

s1.display()
s2.display()
s3.display()
```
